```python
# kNN
from sklearn.neighbors import KNeighborsClassifier

X = filtered_data[["hit_n", "Drums_Energy", "Drums_Complexity", "Variety_of_musical_instruments", "Mixing_Quality", "Harmonic_Richness", "Mixing_Character", "Emotional_Intensity", "track_number"]].values
y = filtered_data["Song_Success"].values
print(X.shape, y.shape)
```

```python
# Деление выборки на обучающую и тестовую
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
# Подбор гиперпараметра
train_accuracies = {}
test_accuracies = {}
neighbors = np.arange(1, 26)
for neighbor in neighbors:
	knn = KNeighborsClassifier(n_neighbors=neighbor)
	knn.fit(X_train, y_train)
	train_accuracies[neighbor] = knn.score(X_train, y_train)
	test_accuracies[neighbor] = knn.score(X_test, y_test)
```

```python
# Вывод графика
plt.figure(figsize=(8, 6))
plt.title("KNN: Varying Number of Neighbors")
plt.plot(neighbors, train_accuracies.values(), label="Training Accuracy")
plt.plot(neighbors, test_accuracies.values(), label="Testing Accuracy")
plt.legend()
plt.xlabel("Number of Neighbors")
plt.ylabel("Accuracy")
plt.show()
```

```python
from sklearn.decomposition import NMF
from sklearn.metrics import mean_squared_error
import numpy as np
import matplotlib.pyplot as plt

# Исходная матрица данных (замените на свои данные)
V = data_list1

# Диапазон числа компонент для тестирования
components = range(1, 12)
errors = []

# Оценка ошибки восстановления для разного числа компонент
for n in components:
    model = NMF(n_components=n, init='random', random_state=42, max_iter=500)
    W = model.fit_transform(V)
    H = model.components_
    reconstruction = np.dot(W, H)
    error = mean_squared_error(V, reconstruction)
    errors.append(error)
```

```python
# Построение графика
plt.plot(components, errors, marker='o')
plt.xticks(components)
plt.xlabel('Number of Components')
plt.ylabel('Reconstruction Error')
plt.title('Reconstruction Error vs Number of Components')
plt.grid()
plt.show()
```

```python
# Применение НМФ
from sklearn.preprocessing import Normalizer
n_components = 3 # Количество компонентов для разложения
model = NMF(n_components=n_components, init='random', random_state=42)

# Применяем NMF на данных
W = model.fit_transform(data_list1) # Матрица весов
H = model.components_ # Матрица компонент

# Печать результатов
print("Матрица W (веса):")
print(W)
print("\nМатрица H (компоненты):")
print(H)

# Восстановленная матрица
X_reconstructed = np.dot(W, H)
print("\nВосстановленная матрица X (приближенная):")
print(X_reconstructed)

# Нормализация признаков (необязательно, но часто полезно для улучшения качества)
normalizer = Normalizer()
X_normalized = normalizer.fit_transform(data_list1)

# Применение НМФ на нормализованных данных
W_normalized = model.fit_transform(X_normalized)
H_normalized = model.components_

print("\nНормализованная матрица W:")
print(W_normalized)
print("\nНормализованная матрица H:")
print(H_normalized)
```

```python
# Тепловая карта вклада каждой переменной в каждый фактор
plt.figure(figsize=(10, 5))
sns.heatmap(H_normalized, annot=True, cmap='viridis', cbar=True)
plt.xlabel('Features (Признаки)')
plt.ylabel('Components (Компоненты)')
plt.title('Вклад признаков в компоненты (матрица H)')
plt.show()
```
