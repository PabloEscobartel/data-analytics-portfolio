```python
import pandas as pd
```

```python
event_windows = [(-2, +2), (-0, +1), (-0, +7)]
estimation_sizes = [90, 180]

results_list = []

# Итерация по всем комбинациям параметров
for event_window in event_windows:
    for estimation_size in estimation_sizes:
        try:
            # Запуск анализа
            release_results = es.Multiple.from_csv(
                path='/content/after_2014_event_weak_experience_mess (1).csv',
                event_study_model=es.Single.market_model,
                event_window=event_window,
                estimation_size=estimation_size,
                buffer_size=14,
                date_format='%d/%m/%Y',
                ignore_errors=True
            )

            # Получение результатов
            result_table = release_results.results(decimals=[3, 5, 3, 5, 2, 2])

            # Добавление последней строки с указанием параметров
            last_row = result_table.iloc[-1].copy()
            last_row['event_window'] = str(event_window)
            last_row['estimation_size'] = estimation_size

            # Добавление строки в список
            results_list.append(last_row)

        except Exception as e:
            print(f"Ошибка для event_window {event_window} и estimation_size {estimation_size}: {e}")

# Конвертация списка строк в DataFrame
results_df = pd.DataFrame(results_list)

# Сохранение результатов в Excel
results_df.to_excel('event_study_results.xlsx', index=False)

print("Результаты сохранены в 'event_study_results.xlsx'.")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from docx import Document
from docx.shared import Inches


# Установка пути к директории с файлами
data_directory = '/content/'  # Путь к директории, где хранятся файлы
output_directory = '/content/results/'  # Путь для сохранения результатов

# Убедимся, что директория для результатов существует
os.makedirs(output_directory, exist_ok=True)

# Получаем список CSV-файлов в директории
csv_files = [f for f in os.listdir(data_directory) if f.endswith('.csv')]

# Комбинации параметров
event_windows = [(-2, +2), (-0, +7)]
estimation_sizes = [180]

# Создаем Word-документ
doc = Document()
doc.add_heading('Анализ Event Study для всех файлов', level=1)

# Обработка каждого файла
for csv_file in csv_files:
    file_path = os.path.join(data_directory, csv_file)
    doc.add_heading(f'Анализ файла: {csv_file}', level=2)

    results_list = []  # Результаты для текущего файла

    for event_window in event_windows:
        for estimation_size in estimation_sizes:
            try:
                # Запуск анализа
                release_results = es.Multiple.from_csv(
                    path=file_path,
                    event_study_model=es.Single.market_model,
                    event_window=event_window,
                    estimation_size=estimation_size,
                    buffer_size=14,
                    date_format='%d/%m/%Y',
                    ignore_errors=True
                )


                # Построение графика
                plt.figure()
                release_results.plot(confidence=0.95)
                plt_path = os.path.join(output_directory, f"plot_{csv_file}_{event_window}_{estimation_size}.png")
                plt.savefig(plt_path)
                plt.close()

                # Добавление описания и графика в Word
                doc.add_heading(f"Event Window: {event_window}, Estimation Size: {estimation_size}", level=3)
                doc.add_paragraph(f"График для файла {csv_file}, окна событий {event_window} и размера оценки {estimation_size}.")
                doc.add_picture(plt_path, width=Inches(5))

            except Exception as e:
                print(f"Ошибка для файла {csv_file}, event_window {event_window} и estimation_size {estimation_size}: {e}")

# Сохранение общего Word-документа
doc_path = os.path.join(output_directory, 'event_study_report.docx')
doc.save(doc_path)

print(f"Все результаты сохранены в директории: {output_directory}")
```
